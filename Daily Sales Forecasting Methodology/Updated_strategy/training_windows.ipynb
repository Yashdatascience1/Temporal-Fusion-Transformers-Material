```python
import pandas as pd
```

```python
def get_windows(start_date, end_date, icl, ocl):
    start_date = pd.Timestamp(start_date)
    end_date = pd.Timestamp(end_date)

    n_days = (end_date - start_date).days + 1
    n_windows = n_days - (icl + ocl) + 1

    rows = []
    for i in range(n_windows):
        input_start  = start_date + pd.Timedelta(days=i)
        input_end    = input_start + pd.Timedelta(days=icl - 1)
        output_start = input_end + pd.Timedelta(days=1)
        output_end   = output_start + pd.Timedelta(days=ocl - 1)
        rows.append({"window": i, "input_start": input_start, "input_end": input_end,
                     "output_start": output_start, "output_end": output_end})

    return pd.DataFrame(rows)
```

```python
get_windows('2023-04-01','2025-12-31',icl=365,ocl=184)
```
