```python
from datetime import datetime
import pandas as pd
from dateutil.relativedelta import relativedelta

def get_tft_sliding_windows_df(start_date_str, end_date_str, input_chunk_len, output_chunk_len, stride_months=1):
    """
    Generates a Pandas DataFrame containing sliding window boundaries 
    for TFT training and evaluation.
    
    Parameters:
    -----------
    start_date_str : str
        The start date of the dataset (e.g., '2023-04-01').
    end_date_str : str
        The maximum available date in the dataset (e.g., '2027-04-01').
    input_chunk_len : int
        Number of months for the historical lookback period.
    output_chunk_len : int
        Number of months for the forecast horizon period.
    stride_months : int, default 1
        How many months to slide the window forward at each step.
        
    Returns:
    --------
    pd.DataFrame
        DataFrame with columns: ['WINDOW_IDX', 'INPUT_START', 'INPUT_END', 'OUTPUT_START', 'OUTPUT_END']
    """
    global_start = datetime.strptime(start_date_str, '%Y-%m-%d')
    global_end = datetime.strptime(end_date_str, '%Y-%m-%d')
    
    total_window_months = input_chunk_len + output_chunk_len
    current_input_start = global_start
    window_idx = 0
    
    # List to collect rows before DataFrame conversion
    window_rows = []
    
    while True:
        # Calculate boundaries (-1 because the start month is inclusive)
        current_input_end = current_input_start + relativedelta(months=input_chunk_len - 1)
        current_output_start = current_input_end + relativedelta(months=1)
        current_output_end = current_input_start + relativedelta(months=total_window_months - 1)
        
        # Break condition: If the forecast window overshoots the maximum data timeline
        if current_output_end > global_end:
            break
            
        window_rows.append({
            "WINDOW_IDX": window_idx,
            "INPUT_START": current_input_start.strftime('%Y-%m-%d'),
            "INPUT_END": current_input_end.strftime('%Y-%m-%d'),
            "OUTPUT_START": current_output_start.strftime('%Y-%m-%d'),
            "OUTPUT_END": current_output_end.strftime('%Y-%m-%d')
        })
        
        # Advance the window
        current_input_start += relativedelta(months=stride_months)
        window_idx += 1
        
    # Convert gathered rows into a clean Pandas DataFrame
    return pd.DataFrame(window_rows)
```

```python
get_tft_sliding_windows_df('2023-04-01','2025-11-01',16,2,1)
```
